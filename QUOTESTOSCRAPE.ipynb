```python
 import urllib.request
from bs4 import BeautifulSoup
import pandas as pd

headers = {'User-Agent': 'Mozilla/5.0'}
all_quotes = []

for a in range(1,11):
    link = 'https://quotes.toscrape.com/page/' + str(a) + '/'
    req = urllib.request.Request(link, headers=headers)
    fhand = urllib.request.urlopen(req)
    soup = BeautifulSoup(fhand, 'html.parser')


    quotes = soup.find_all('div', class_='quote')
    for q in quotes:
        quote2 = q.find('span', class_='text').get_text()
        author = q.find('small', class_='author').get_text()
        tags = [t.get_text() for t in q.find_all('a', class_='tag')]



        all_quotes.append({
            "Quote": quote2,
            "Author": author,
            "tags": tags})


df = pd.DataFrame(all_quotes)
df.to_excel("Quotes.xlsx", index=False)
print("Scraped Author Quotes saved to Quotes.xlsx")

    
```
