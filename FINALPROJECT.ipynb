```python
import urllib.request
from bs4 import BeautifulSoup
import pandas as pd
import time

headers = {'User-Agent': 'Mozilla/5.0'}
metacritic_stats = []

# Loop through all 42 pages
for a in range(1, 43):
    print(f"Scraping listing page {a} of 42...")
    list_url = f'https://www.metacritic.com/browse/movie/?releaseYearMin=1910&releaseYearMax=2025&page={a}'
    
    try:
        req = urllib.request.Request(list_url, headers=headers)
        fhand = urllib.request.urlopen(req)
        soup = BeautifulSoup(fhand, 'html.parser')
    except Exception as e:
        print(f"Failed to load listing page {a}: {e}")
        continue

    # Find all movie cards
    movies = soup.find_all('a', class_="c-finderProductCard_container")
    for e in movies:
        title_tag = e.find('h3')
        movie_name = title_tag.get_text(strip=True) if title_tag else "N/A"

        href = e.get('href')
        movie_link = f"https://www.metacritic.com{href}" if href else "N/A"

        # Visit movie detail page
        try:
            req2 = urllib.request.Request(movie_link, headers=headers)
            fhand2 = urllib.request.urlopen(req2)
            soup2 = BeautifulSoup(fhand2, 'html.parser')
        except Exception as e:
            print(f"Failed to load detail page for {movie_name}: {e}")
            continue

        # Extract fields from detail page
        def get_field(label):
            tag = soup2.find('span', string=label)
            return tag.find_next('span').get_text(strip=True) if tag else "N/A"

        meta_score_tag = soup2.find('span', class_="c-siteReviewScore")
        meta_score = meta_score_tag.get_text(strip=True) if meta_score_tag else "N/A"

        user_score_tag = soup2.find('span', class_="c-siteReviewScore_user")
        user_score = user_score_tag.get_text(strip=True) if user_score_tag else "N/A"

        release_date = get_field("Release Date")
        genre = get_field("Genre(s)")
        director = get_field("Director")
        actors = get_field("Starring")

        # Save the row
        metacritic_stats.append({
            "Title": movie_name,
            "Movie Link": movie_link,
            "Meta Score": meta_score,
            "User Score": user_score,
            "Release Date": release_date,
            "Genre": genre,
            "Director": director,
            "Actors": actors
        })

        time.sleep(0.5)  # polite delay between movie requests

# Save to Excel
df = pd.DataFrame(metacritic_stats)
df.to_excel("metacritic.xlsx", index=False)
print("Scraped Metacritic movie data saved to metacritic.xlsx")
```

```python
import urllib.request
import pandas as pd
from bs4 import BeautifulSoup

headers = {'User-Agent': 'Mozilla/5.0 (Macintosh; Intel Mac OS X 10_13_6) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/70.0.3538.77 Safari/537.36'}

movies_data = []
n = 1

for a in range(1, 43):
    link = f'https://www.metacritic.com/browse/movie/?releaseYearMin=1910&releaseYearMax=2025&page={a}'
    req = urllib.request.Request(link, headers=headers)
    fhand = urllib.request.urlopen(req)
    soup = BeautifulSoup(fhand, 'html.parser')

    movies = soup.find_all('a', class_="c-finderProductCard_container g-color-gray80 u-grid")
    
    for e in movies:
        movie_info = {}
        
        # Movie name & link
        movie_tag = e.find('h3')
        movie_info['Title'] = movie_tag.get_text().strip()
        href = e.get('href')
        movie_info['Link'] = 'https://www.metacritic.com' + href
        
        # Go to movie page
        req = urllib.request.Request(movie_info['Link'], headers=headers)
        fhand = urllib.request.urlopen(req)
        movie_soup = BeautifulSoup(fhand, 'html.parser')
        
        # Directors
        try:
            directors_row = movie_soup.find('div', class_="c-crewList g-inner-spacing-bottom-small c-productDetails_staff_directors")
            directors = [i.get_text().strip() for i in directors_row.find_all('a')]
            movie_info['Directors'] = ', '.join(directors)
        except:
            movie_info['Directors'] = None
        
        # Summary
        try:
            summary_text = movie_soup.find('span', class_="c-productDetails_description g-text-xsmall")
            movie_info['Summary'] = summary_text.get_text().strip()
        except:
            movie_info['Summary'] = None
        
        # Actors
        try:
            actors_row = movie_soup.find('div', class_="c-globalCarousel_content c-globalCarousel_content-scrollable c-globalCarousel_content-scrollable_mobile-gap-small")
            actors = [v.get_text().strip() for v in actors_row.find_all('h3')]
            movie_info['Actors'] = ', '.join(actors)
        except:
            movie_info['Actors'] = None
        
        # Genres
        try:
            genres = movie_soup.find('ul', class_="c-genreList u-flexbox g-inner-spacing-top-medium")
            genre_list = [k.get_text().strip() for k in genres.find_all('a')]
            movie_info['Genres'] = ', '.join(genre_list)
        except:
            movie_info['Genres'] = None
        
        # User score
        try:
            user_score = movie_soup.find('div', class_="c-siteReviewScore_user")
            movie_info['User Score'] = user_score.get_text().strip()
        except:
            movie_info['User Score'] = None
        
        # Meta score
        try:
            meta_score = movie_soup.find('div', class_="c-siteReviewScore g-text-bold")
            movie_info['Metascore'] = meta_score.get_text().strip()
        except:
            movie_info['Metascore'] = None
        
        # Release date
        try:
            release_time = e.find('span', class_="u-text-uppercase")
            movie_info['Release Date'] = release_time.get_text().strip()
        except:
            movie_info['Release Date'] = None
        
        # Append movie to list
        movies_data.append(movie_info)
        
        print(f"Processed movie #{n}: {movie_info['Title']}")
        
        n += 1
        if n > 1000:
            break
    if n > 1000:
        break

# Save to Excel
df = pd.DataFrame(movies_data)
df.to_excel("metacritic_movies.xlsx", index=False)
print("✅ Data saved to metacritic_movies.xlsx")
```
